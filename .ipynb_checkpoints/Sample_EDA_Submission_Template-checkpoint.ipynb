# **Problem Statement**


**Business Context**

Paisabazaar is a financial services company that assists customers in finding and applying for various banking and credit products. An integral part of their service is assessing the creditworthiness of individuals, which is crucial for both loan approval and risk management. The credit score of a person is a significant metric used by financial institutions to determine the likelihood that an individual will repay their loans or credit balances. Accurate classification of credit scores can help Paisabazaar enhance their credit assessment processes, reduce the risk of loan defaults, and offer personalized financial advice to their customers.

In this context, analyzing and classifying credit scores based on customer data can improve decision-making processes and contribute to better financial product recommendations. This case study aims to develop a model that predicts the credit score of individuals based on various features, such as income, credit card usage, and payment behavior.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code
income_by_score = data.groupby('Credit_Score')['Annual_Income'].mean().reindex(['Poor','Standard','Good'])

plt.figure(figsize=(8,6))
sns.barplot(
    x=income_by_score.index,
    y=income_by_score.values,
    hue=income_by_score.index,  # use x variable as hue
    palette='viridis',
    legend=False
)
plt.title('Average Annual Income by Credit Score Category')
plt.xlabel('Credit Score Category')
plt.ylabel('Average Annual Income')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
````

##### 1. Why did you pick the specific chart?

Credit_Score is categorical while Annual_Income is numerical. So Bar plot of average annual income across different categories in Credit_Score can show us the trend of annual income vs credit score.

##### 2. What is/are the insight(s) found from the chart?

From the plot we can see that as we go from poor to good the average annual income is increasing. Therefore our assumption regarding positive correlation b/w Annual_Income and Credit_Score is correct.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The insight observed is helpful in creating positive impact. It allows us to have a positive weightage on a parameter ie annual income in estimating the credit score.

#### Chart - 2

````python
# Chart - 2 visualization code
# Compute group means
group_means = data.groupby('Credit_Score')[['Credit_Utilization_Ratio', 'Num_of_Loan']].mean().reindex(['Poor', 'Standard', 'Good'])

# Normalize each column (so they’re comparable on 0–1 scale)
normalized_means = group_means / group_means.max()

# Plot normalized means
normalized_means.plot(
    kind='bar',
    figsize=(10,8),
    color=['#4C72B0', '#55A868']
)
plt.title('Normalized Average of Credit Utilization Ratio and Number of Loans by Credit Score')
plt.xlabel('Credit Score Category')
plt.ylabel('Normalized Average Value (0–1 scale)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=0)
plt.legend(['Credit Utilization Ratio', 'Number of Loans'])
plt.show()
````

````python
# Encode Credit_Score to numeric scale for correlation
score_map = {'Poor': 1, 'Standard': 2, 'Good': 3}
data['Credit_Score_Num'] = data['Credit_Score'].map(score_map)

# Compute correlation values
corr_util = data['Credit_Score_Num'].corr(data['Credit_Utilization_Ratio'])
corr_loan = data['Credit_Score_Num'].corr(data['Num_of_Loan'])

print("Correlation with Credit Utilization Ratio:", corr_util)
print("Correlation with Number of Loans:", corr_loan)
````

##### 1. Why did you pick the specific chart?

As mentioned before with the categorical data type of credit score relationship is better recognizable with bar plots.

Also since we are comparing the relation strength of two numerical variables this time i normalized both the means of the variables, so they are easily comparable.

##### 2. What is/are the insight(s) found from the chart?

Credit Utilization Ratio has a little positive correlation where as No of loans has very prominent negative correlation with credit score.

The correlation calculation also tells the same.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

The insight obviously adds a new reliable parameter no of loans which has negative weightage on calculating the credit score.

#### Chart - 3

````python
# Chart - 3 visualization code
# Define what counts as an adequate credit utilization ratio (e.g. ≤ 30%)
adequate_ratio = 30

# Create a categorical column for utilization status
data['Managed_Credit_Utilization_Ratio'] = data['Credit_Utilization_Ratio'] <= adequate_ratio

grouped_data = data.groupby(['Credit_Score', 'Managed_Credit_Utilization_Ratio'])['Num_Credit_Card'].mean().reset_index()

# Plot grouped bar chart
plt.figure(figsize=(8, 6))
sns.barplot(x='Credit_Score', y='Num_Credit_Card', hue='Managed_Credit_Utilization_Ratio', data=grouped_data, order=['Poor', 'Standard', 'Good'])

plt.show()
````

##### 1. Why did you pick the specific chart?

It is clear about the relationship of number of credit cards with credit_score for the same reason as before.

##### 2. What is/are the insight(s) found from the chart?

There is a negative correlation b/w number of credit cards and credit score.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

With this insight we now can include number of credit card with a negative weightage to estimate credit score.

##### 1. Why did you pick the specific chart?

Same reason as before.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Number of delayed payments can be considered a parameter for finding credit score with negative weightage.

#### Chart - 5

````python
# Chart - 5 visualization code
def agegroup(x):
  if x < 20:
    return '10-20'
  if x < 30:
    return '20-30'
  if x < 40:
    return '30-40'
  if x < 50:
    return '40-50'
  if x < 60:
    return '50-60'
  return '60+'

data['Age_Group'] = data['Age'].apply(agegroup)

grouped_data = data[['Customer_ID', 'Age_Group', 'Credit_Score']].drop_duplicates().groupby(['Age_Group', 'Credit_Score']).count().reset_index()

# Plot grouped bar chart
plt.figure(figsize=(10, 6))

poor_counts = grouped_data[grouped_data['Credit_Score'] == 'Poor'].groupby('Age_Group')['Customer_ID'].sum()
standard_counts = grouped_data[grouped_data['Credit_Score'] == 'Standard'].groupby('Age_Group')['Customer_ID'].sum()
good_counts = grouped_data[grouped_data['Credit_Score'] == 'Good'].groupby('Age_Group')['Customer_ID'].sum()

plt.bar(grouped_data['Age_Group'].unique(), poor_counts, label='Poor', color='#2d6fd2')
plt.bar(grouped_data['Age_Group'].unique(), standard_counts, bottom=poor_counts, label='Standard', color='#e481c3')
plt.bar(grouped_data['Age_Group'].unique(), good_counts, bottom=poor_counts+standard_counts,label='Good', color='#d26f2d')

plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.title('Number of Customers by Age Group and Credit Score')
plt.legend()

plt.show()
````

##### 1. Why did you pick the specific chart?

Like Credit_Score Age Group's also categorical and we are having other variable that is count of customers. 
Since each customer has a 8 records corresponding to 8 months i had to first remove duplicates based on the target columns so that there is only one record for each customer. Bar graph is the best for categorical vs numerical but stacked bar graph is best if we need to represent distribution of other variable.

##### 2. What is/are the insight(s) found from the chart?

Regarding credit score distribution, In all age groups the dominant credit score is 'Standard' and least dominant is 'Good'. Other key things to note is that in age group '10-20' the 'Good' credit score is scarce where as in '50-60' the 'Poor' credit score is scarce.

Other probably not so relevant thing is most of the customers are in age groups '20-40'.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

In age group '10-20' the relative ratio of 'Poor' credit score count to the whole age group head count is the highest. So it could have a negative feedback to the age group's credit score.

Its Numerical(Month) vs Numerical(Count) so line plot is the best.

##### 2. What is/are the insight(s) found from the chart?

In Month april there is a dip in Standard customers where as there is a spike in Good customers. The Poor customer count is nearly a straight line throuout all the months. The Customers count distribution through out all months have the order Standard > Poor > Good.

##### 3. Will the gained insights help creating a positive business impact?

The dip and spike in Standard and Good customers count can possibly insinuate that in the month april some customers who have standard credit score might perform well to promote to good.
